Finding the headers in the csv file

```python
import pandas as pd

# Replace 'your_file.csv' with your actual file name or path
file_path = 'listings.csv'

# Read just the header row (nrows=0 avoids loading the whole file into memory)
df = pd.read_csv(file_path, nrows=0)
headings = df.columns.tolist()

# Print each heading clearly on a new line with its column number
for i, heading in enumerate(headings, 1):
    print(f"{i}: {heading}")
```
